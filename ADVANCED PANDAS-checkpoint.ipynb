```python
# sns.get_dataset_names()
df=sns.load_dataset("titanic")
df
```

```python
a=df.loc[:,['pclass','who','fare']]
a
```

```python
data=sns.load_dataset('diamonds')
data
```

```python
data.loc[(data.color=="D")&(data.cut=='Ideal'),"price"].mean()
```

```python
pd.pivot_table(data=data,index=["color"],columns=["cut"],values=["price"],margins=False).T
```

```python
#avg price in each clarity group by cut()
data.groupby(by=['color',"cut"],observed=True)['price'].mean()
```

```python
pd.pivot_table(data=data,index=['cut'],columns=data['color'],values=['color'],aggfunc='count',observed=False)
```

```python
pd.crosstab(index=data['cut'],columns=data['color'],values=data['color'],aggfunc='mean')
```
